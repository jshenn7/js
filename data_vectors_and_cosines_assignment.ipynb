```python
%pip install seaborn

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from numpy import linalg

from scipy import stats
from scipy.stats import norm, probplot
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

%matplotlib inline

#Part 1:

age = [13, 15, 16, 16, 19, 20, 20, 21, 22, 22, 25, 25, 25, 25, 30, 33, 33, 35, 35, 35, 35, 36, 40, 45, 46, 52, 70]

np.mean(age)
np.median(age)
#29.962962962962962
#25.0

stats.mode(age)
plt.hist(age)
np.percentile(age, [25, 75])
#Modes: 25, 35; bimodal
#Q1: 20.5, Q3: 35
#Min: 13, Q1: 20.5, Median: 25.0, Q3: 35.0, Max: 70

plt.boxplot(age)
plt.show()
#Outlier at 70, median at 25, IQR from 20.5 to 35, data is skewed right towards the upper end

#Part 2:

iris = load_iris()
iris_df = pd.DataFrame(data= np.c_[iris['data'], iris['target']], columns= iris['feature_names'] + ['target'])
iris_df.head()

#For sepal length, sepal width, petal length, and petal width:
    #Continuous and numeric
    #Order? - Yes
    #Interval? - Yes
    #Zero Point - Yes
    #Ratio
#For target
    #Discrete
    #No order, interval, or zero point
    #Nominal

sns.distplot(iris_df["sepal length (cm)"])
#One peak, roughly symmetric, normal distribution
sns.distplot(iris_df["sepal width (cm)"])
#One peak, roughly symmetric, normal distribution
sns.distplot(iris_df["petal length (cm)"])
#Two peaks, not symmetric, odd distribution, gap in data
sns.distplot(iris_df["petal width (cm)"])
#Two peaks, not symmetric, odd distribution, gap in data

probplot(iris_df["sepal length (cm)"], dist="norm", plot=plt)
probplot(iris_df["sepal width (cm)"], dist="norm", plot=plt)
probplot(iris_df["petal length (cm)"], dist="norm", plot=plt)
probplot(iris_df["petal width (cm)"], dist="norm", plot=plt)
#Petal length and width are not normal

sns.boxplot(x="target", y="sepal length (cm)", data=iris_df)
#Not much of a constant variance across targets
#Somewhat symmetrical, except for 2.0 which is slightly positively skewed
#The means are different across targets
```
